```python
# 5. Mutable vs Immutable
def modify_values(a_list, a_tuple):
    a_list[0] = 100
    try:
        a_tuple[0] = 100
    except TypeError:
        print("Tuples are immutable.")

my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
modify_values(my_list, my_tuple)
print("Modified List:", my_list)
print("Original Tuple:", my_tuple)
```

```python
# 7. Logical operators
x = 10
y = 5
print(x > 5 and y < 10)  # True
print(x < 5 or y < 10)   # True
print(not(x < 5))        # True
```

```python
# 8. List vs Tuple modification
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

my_list[1] = 99  # Allowed
print("Modified List:", my_list)

try:
    my_tuple[1] = 99  # Will cause an error
except TypeError:
    print("Cannot modify a tuple!")
```

```python
# 13. Sum of even numbers between 1 and 50
sum_even = sum(i for i in range(1, 51) if i % 2 == 0)
print("Sum of even numbers from 1 to 50:", sum_even)
```

```python
# 14. Reverse a string using while
text = input("Enter a string: ")
reversed_str = ""
i = len(text) - 1
while i >= 0:
    reversed_str += text[i]
    i -= 1
print("Reversed String:", reversed_str)
```

```python
# 15. Factorial using while
n = int(input("Enter a number: "))
fact = 1
i = 1
while i <= n:
    fact *= i
    i += 1
print(f"Factorial of {n} is {fact}")
```
